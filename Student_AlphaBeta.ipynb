```python
import math

HUMAN, AI, EMPTY = 'X', 'O', ' '
board = [EMPTY] * 9
buttons = []

# PROTECTED SECTION - DO NOT MODIFY BELOW THIS LINE
def winner():
    wins = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6),
    ]
    for a, b, c in wins:
        if board[a] == board[b] == board[c] != EMPTY:
            return board[a]
    return 'DRAW' if EMPTY not in board else None

def avail():
    return [i for i, v in enumerate(board) if v == EMPTY]
# PROTECTED SECTION END
```

```python
import math
from pyvirtualdisplay import Display

# Start headless virtual display to prevent TclError in Colab
vdisplay = Display(visible=0, size=(800, 600))
vdisplay.start()




def alphabeta(depth, alpha, beta, maximizingPlayer):
    """
    Implement the Alpha-Beta Pruning algorithm.

    Parameters
    ----------
    depth : int
        Current depth of the search tree.

    alpha : float
        Best value that the maximizing player can guarantee.

    beta : float
        Best value that the minimizing player can guarantee.

    maximizingPlayer : bool
        True  -> AI's turn (Maximizer)
        False -> Human's turn (Minimizer)

    Returns
    -------
    int
        The evaluation score of the current board state.

    Utility: Scoring Rules
    -------------
    AI wins    : 10 - depth
    Human wins : depth - 10
    Draw       : 0

    Notes
    -----
    - Use the global `board` variable.
    - Use `winner()` to detect terminal states.
    - Use `avail()` to obtain all possible moves.
    - Apply Alpha-Beta Pruning exactly as discussed in class.
    - Do NOT modify any other part of the program.
    """

    # Step 1:
    # Check whether the current board is a terminal state.
    # If the game has ended, return the appropriate score.

    # TODO WRITE CODE HERE

    if winner()=="X":
        return depth-10
    elif winner()=="O":
        return 10-depth
    elif winner()=="DRAW":
        return 0


    # Step 2:
    # If it is the AI's turn (maximizing player):
    #   - Initialize the best value.
    #   - Try every available move.
    #   - Recursively evaluate each move.
    #   - Update alpha.
    #   - Apply Alpha-Beta Pruning when possible.


    if maximizingPlayer:
        maxEval=float("-inf")
        moves=avail()
        for move in moves:
            board[move]=AI
            alpha=max(alpha,alphabeta(depth+1,alpha,beta,False))
            if alpha>=beta:
                return alpha
        maxEval=max(maxEval,alpha)
        return maxEval


        # TODO WRITE CODE HERE


    # Step 3:
    # If it is the Human's turn (minimizing player):
    #   - Initialize the best value.
    #   - Try every available move.
    #   - Recursively evaluate each move.
    #   - Update beta.
    #   - Apply Alpha-Beta Pruning when possible.

    else:
        minEval=float("inf")
        moves=avail()
        for move in moves:
            board[move]=HUMAN
            beta=min(beta,alphabeta(depth+1,alpha,beta,True))
            if alpha>=beta:
                return beta
        minEval=min(minEval,beta)
        return minEval


        # TODO WRITE CODE HERE


    # Return the computed evaluation value.
    raise NotImplementedError("Implement the Alpha-Beta Pruning algorithm.")
```

```python
# PROTECTED SECTION - DO NOT MODIFY BELOW THIS LINE



def best_move():
    best = -math.inf
    move = None
    for m in avail():
        board[m] = AI
        score = alphabeta(0, -math.inf, math.inf, False)
        board[m] = EMPTY
        if score > best:
            best = score
            move = m
    return move

def end():
    r = winner()
    if not r:
        return False
    try:
        from tkinter import messagebox
        messagebox.showinfo('Game Over', 'Draw!' if r == 'DRAW' else f'{r} wins!')
    except Exception:
        print('Game Over:', 'Draw!' if r == 'DRAW' else f'{r} wins!')
    reset()
    return True

def play(i):
    if board[i] != EMPTY:
        return
    board[i] = HUMAN
    buttons[i]['text'] = HUMAN
    if end():
        return
    m = best_move()
    if m is not None:
        board[m] = AI
        buttons[m]['text'] = AI
    end()

def reset():
    global board
    board = [EMPTY] * 9
    for b in buttons:
        b.config(text='')

def create_gui():
    import tkinter as tk
    root = tk.Tk()
    root.title('Alpha-Beta')
    buttons.clear()
    for i in range(9):
        bt = tk.Button(root, text='', width=6, height=3, font=('Arial', 20), command=lambda i=i: play(i))
        bt.grid(row=i // 3, column=i % 3)
        buttons.append(bt)
    tk.Button(root, text='Reset', command=reset).grid(row=3, column=0, columnspan=3, sticky='ew')
    root.mainloop()

def main():
    create_gui()

if __name__ == '__main__':
    main()
# PROTECTED SECTION END
```
